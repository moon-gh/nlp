Corpus(말뭉치) : 분석에 활용하기 위한 자연어 data

corpus를 분석에 활용하려면 먼저 의미있는 작은 단위(Token)로 나누어야 함.

Tokenization(토큰화) : 1개의 Corpus를 여러 개의 토큰으로 나누는 과정 

```python
from nltk.tokenize import word_tokenize
```

nltk에서 제공하는 토큰화 모듈 punkt

punkt는 마침표 or 약어(Mr.. Dr.)와 같은
특별한 언어적 특성을 고려하여 토큰화를 할 수 있게 해주는 모듈

```python
import nltk
nltk.download('punkt')
```

nltk의 word_tokenizer()
: corpus를 parameter로 넘겨서 토큰화된 단어 list를 반환하는 함수

```python
text = "Although it's not a happily-ever-after ending, it is very realistic."

tokenized_words = word_tokenize(text)

print(tokenized_words)
```
